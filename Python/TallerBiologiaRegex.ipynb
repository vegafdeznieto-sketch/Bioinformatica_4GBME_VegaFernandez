##### ¿Puedes hacer que devuelva dos decimales? ¿Y que lidie con mayúsculas y minúsculas?

```python
def porcentajeA(secuencia):
    secuencia = secuencia.upper()
    porcentaje = (secuencia.count("A")/len(secuencia))*100.00
    porcentaje_redondeado = round(porcentaje,2)
    print(porcentaje_redondeado)
```
